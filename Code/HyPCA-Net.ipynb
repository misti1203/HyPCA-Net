```python
import numpy as np
import cv2

def rotate_image(image, angle):
    """
    Rotate the image by the specified angle.
    """
    center = tuple(np.array(image.shape[1::-1]) / 2)
    rotation_matrix = cv2.getRotationMatrix2D(center, angle, 1.0)
    rotated_image = cv2.warpAffine(image, rotation_matrix, image.shape[1::-1], flags=cv2.INTER_LINEAR)
    return rotated_image

def translate_image(image, tx, ty):
    """
    Translate the image by the specified translation parameters.
    """
    translation_matrix = np.float32([[1, 0, tx], [0, 1, ty]])
    translated_image = cv2.warpAffine(image, translation_matrix, image.shape[1::-1])
    return translated_image

def apply_gaussian_blur(image, kernel_size=3):
    """
    Apply Gaussian Blur to the image to reduce noise and improve generalization.
    """
    blurred_image = cv2.GaussianBlur(image, (kernel_size, kernel_size), 0)
    return blurred_image

# Augmentation parameters
rotation_angles = [20]
translations = [(5, 5)]
kernel_sizes = [3]  # Gaussian Blur kernel sizes

augmented_X_train = []
augmented_y_train = []

for image, label in zip(X_train_s, y_train_s):
    # Augment with rotations
    for angle in rotation_angles:
        rotated_image = rotate_image(image, angle)
        augmented_X_train.append(rotated_image)
        augmented_y_train.append(label)

    # Augment with translations
    for tx, ty in translations:
        translated_image = translate_image(image, tx, ty)
        augmented_X_train.append(translated_image)
        augmented_y_train.append(label)

    # Augment with Gaussian Blur
    for kernel_size in kernel_sizes:
        blurred_image = apply_gaussian_blur(image, kernel_size)
        augmented_X_train.append(blurred_image)
        augmented_y_train.append(label)

# Convert lists to numpy arrays
augmented_X_train = np.array(augmented_X_train)
augmented_y_train = np.array(augmented_y_train)

# Shuffle the data
shuffle_indices = np.random.permutation(len(augmented_X_train))
augmented_X_train = augmented_X_train[shuffle_indices]
augmented_y_train = augmented_y_train[shuffle_indices]
augmented_X_train.shape, augmented_y_train.shape
```

```python
import tensorflow as tf
from tensorflow.keras import layers

def channel_attention_module(conv_layer):  ## CHIA Channel Holistic Information Attention
    def apply(x):
        avg_pool = tf.reduce_mean(x, axis=[1, 2], keepdims=True)
        max_pool = tf.reduce_max(x, axis=[1, 2], keepdims=True)
        min_pool = tf.reduce_min(x, axis=[1, 2], keepdims=True)
        sum_pool = tf.reduce_sum(x, axis=[1, 2], keepdims=True)
        pooled = avg_pool + max_pool + min_pool + sum_pool
        gate = conv_layer(pooled)
        #return x * gate
        return gate
    return apply

def spatial_attention_module(conv_layer):  ## SHIA Spatial Holistic Information Attention
    def apply(x):
        avg_pool = tf.reduce_mean(x, axis=-1, keepdims=True)
        max_pool = tf.reduce_max(x, axis=-1, keepdims=True)
        min_pool = tf.reduce_min(x, axis=-1, keepdims=True)
        sum_pool = tf.reduce_sum(x, axis=-1, keepdims=True)
        concat = avg_pool + max_pool + min_pool + sum_pool #tf.concat([avg_pool, max_pool, min_pool, sum_pool], axis=-1)
        spatial_map = conv_layer(concat)
        #return x * spatial_map
        return spatial_map
    return apply



class ChannelSpatialAttention(tf.keras.layers.Layer):  ## Spatial-Channel Adaptive Learning Attention (SCALA)
    def __init__(self):
        super(ChannelSpatialAttention, self).__init__()
        # We initialize Conv layers as None and define them in build()
        #self.channel_conv = None
        #self.spatial_conv = None
        #self.channel_attention = None
        #self.spatial_attention = None

        # Learnable fusion weights
        self.weight1 = self.add_weight(shape=(1, 1, 1, 1), initializer=tf.keras.initializers.HeNormal(), trainable=True, name='w_channel')
        self.weight2 = self.add_weight(shape=(1, 1, 1, 1), initializer=tf.keras.initializers.HeNormal(), trainable=True, name='w_spatial')
        self.weight3 = self.add_weight(shape=(1, 1, 1, 1), initializer=tf.keras.initializers.HeNormal(), trainable=True, name='w_attention')

    def build(self, input_shape):
        channels = input_shape[-1]
        self.channel_conv = layers.Dense(channels, activation='sigmoid')
        self.spatial_conv = layers.Conv2D(1, kernel_size=7, padding='same', activation='sigmoid')

        # Wrap the functions now that layers are available
        self.channel_attention = channel_attention_module(self.channel_conv)
        self.spatial_attention = spatial_attention_module(self.spatial_conv)

    def call(self, x):
        channel_info = self.channel_attention(x)
        spatial_info = self.spatial_attention(x)

        #channel_info = self.weight1 * channel_info
        #spatial_info = self.weight2 * spatial_info

        attention_map = tf.sigmoid(channel_info + spatial_info)

        x_out = x * attention_map #* self.weight3
    
        return x_out
```

```python
def multi_kernel_groupwise_conv1(x, filters, groups=4, strides=2):
    """ Advanced Multi-Kernel Groupwise Convolution Block (Without Attention, Optimized) """

    attention = ChannelSpatialAttention()
    
    # Multi-Kernel Parallel Convolutions with Different Receptive Fields
    conv1x1 = layers.Conv2D(filters // 4, kernel_size=1, padding="same")(x)
    conv3x3 = layers.DepthwiseConv2D(kernel_size=3, padding="same")(x)
    conv5x5 = layers.DepthwiseConv2D(kernel_size=5, padding="same")(x)
    conv_dilated = layers.DepthwiseConv2D(kernel_size=3, dilation_rate=2, padding="same")(x)

    # Feature Fusion via Concatenation
    x1 = layers.Concatenate()([conv1x1, conv3x3, conv5x5, conv_dilated])

    # Channel Shuffle for Better Feature Mixing
    def channel_shuffle(x, groups):
        batch, height, width, channels = tf.unstack(tf.shape(x))
        x = tf.reshape(x, [-1, height, width, groups, channels // groups])
        x = tf.transpose(x, [0, 1, 2, 4, 3])
        x = tf.reshape(x, [-1, height, width, channels])
        return x

    x1 = layers.Lambda(lambda x: channel_shuffle(x, groups))(x1)

    # Depthwise + Grouped Convolutions Hybrid
    x1 = layers.DepthwiseConv2D(kernel_size=3, padding="same")(x1)
    x1 = layers.Conv2D(filters, kernel_size=1, groups= groups, padding="same")(x1)

    # Downsampling x1
    x1 = layers.DepthwiseConv2D(kernel_size=3, strides=strides, padding="same")(x1)

    x1 = attention(x1)

    

    # **Fix: Ensure x has the same number of channels as x1**

    x = layers.Conv2D(filters, kernel_size=1, groups= groups, padding="same")(x)
    x = layers.DepthwiseConv2D(kernel_size=3, strides=strides, padding="same")(x)
    x = layers.Conv2D(filters, kernel_size=1, groups= groups, padding="same")(x)

    # Residual Connection
    x = layers.Add()([x, x1])
    x = layers.Activation("relu")(x)
    
    return x
```

```python
import tensorflow as tf
import tensorflow_probability as tfp
from tensorflow.keras import Model, Input
from tensorflow.keras.layers import (
    Layer, Dense, LayerNormalization,
    Conv2D, BatchNormalization, Add, ReLU,
    Dropout, GlobalAveragePooling2D, GlobalMaxPooling2D, MaxPooling2D
)

# ----------------------------------------------------
# 1. NODEAttentionLayer with adaptive Dormand–Prince
# ----------------------------------------------------
import tensorflow as tf
from tensorflow.keras.layers import Layer, Dense, LayerNormalization
from tensorflow.keras import Model, Input
from tensorflow.keras.layers import (
    Conv2D, BatchNormalization, Add, ReLU,
    Dropout, GlobalAveragePooling2D, GlobalMaxPooling2D, MaxPooling2D
)

# ----------------------------------------------------
# 1. NODEAttentionLayer with one-step Euler integrator
# ----------------------------------------------------
class NODEAttentionLayer(Layer):
    def __init__(self, dim, S_max: int = 4, se_ratio: int = 16, use_scale=True, **kwargs):

        '''dim,
        num_heads: int = 4,
        S_max: int = 4,
        head_skip_thresh: float = 0.1,
        solver_gate: bool = True,
        spatial_adapt: bool = True,
        complexity_loss_weight: float = 1e-4,
        se_ratio: int = 16,'''
        
        """
        dim: feature dimension
        """
        super().__init__(**kwargs)
        self.dim = dim
        self.S_max = S_max
        self.use_scale = use_scale
        
        bottleneck = max(1, dim // 4)
        
        # small MLP for fθ
        self.mlp = tf.keras.Sequential([
            Dense(bottleneck, activation='relu'),
            Dense(dim, activation='sigmoid'),
        ], name="f_theta_mlp")

        # time-head: predicts integration horizon T ∈ (0, 1]
        self.time_head = tf.keras.Sequential([
            Dense(bottleneck, activation='relu'),
            Dense(1, activation='sigmoid'),  # outputs in (0,1)
        ], name="time_head")

        # gating between h0 and h(T)
        self.gate = Dense(dim, activation='sigmoid', name="ode_gate")

        self.step_head = Dense(self.S_max, name="step_head_logits")


    
        reduced_dim = max(1, dim // se_ratio)
        self.se_reduce = Dense(reduced_dim // 4, activation="relu", name="se_reduce")
        self.se_expand = Dense(dim, activation="sigmoid", name="se_expand")

        if self.use_scale:
            self.global_scale = self.add_weight(shape=(1, 1, 1, 1), initializer='ones', trainable=True, name='global_scale')
            self.local_scale = self.add_weight(shape=(1, 1, 1, self.dim), initializer='ones', trainable=True, name='local_scale')
        



    def call(self, h0):
        """
        h0: Tensor of shape (batch_seq, seq_len, dim)
        """
        # predict per-window horizon T in (0,1]
        # reshape so we can broadcast over seq_len & dim
        
        
        # 1) Continuous-depth step prediction
        mean_vec = tf.reduce_mean(h0, axis=1)
        logits = self.step_head(mean_vec)
        probs = tf.nn.softmax(logits, axis=-1)
        steps = tf.reduce_sum(
            probs * tf.cast(tf.range(1, self.S_max+1), probs.dtype),
            axis=-1, keepdims=True
        )
        dt = tf.reshape(
                self.time_head(tf.reduce_mean(h0, axis=1)),  # (batch_seq,1)
                [-1, 1, 1]
            )                                                                        #1.0 / tf.reshape(steps, [batch,1,1,1])

        def euler_step(h0):
            T = tf.reshape(
                self.time_head(tf.reduce_mean(h0, axis=1)),  # (batch_seq,1)
                [-1, 1, 1]
            )  # now (batch_seq, 1, 1)
    
            # one-step Euler: hT = h0 + T * fθ(h0)
            dh = self.mlp(h0)
            hT = h0 + T * dh
            return hT

        #def euler_step(x): return x + dt * self.mlp(x)
        
        def rk2_step(x):
            k1 = self.mlp(x)
            mid = x + (dt/2) * k1
            k2 = self.mlp(mid)
            return x + dt * k2


        '''def euler_step(x):
            return x + dt * self.mlp(x)'''


        '''def heun_step(x):
            k1 = self.mlp(x)
            k2 = self.mlp(x + dt * k1)
            return x + dt * 0.5 * (k1 + k2)'''

        #mean_vec = tf.reduce_mean(h0, axis=1)
        
        for _ in range(self.S_max): 
            h_e = euler_step(h0)
            h_r = rk2_step(h0)
            #h_h = heun_step(h0)
            
            hT = h_r + h_e #+ h_h

        #hT = hT + h0

        

        

        # gated residual blending
        gate_vals = self.gate(h0)  # (batch_seq, seq_len, dim)
        out1 = gate_vals * hT + (1.0 - gate_vals) * h0

        out = out1

        # ─── 8) Squeeze-and-Excitation channel attention ───────────
        #    Squeeze: global average pool over tokens
        se1 = tf.reduce_mean(out, axis=1, keepdims=True)  # (B,1,dim)
        se2 = tf.reduce_max(out, axis=1, keepdims=True)  # (B,1,dim)
        se3 = tf.reduce_min(out, axis=1, keepdims=True)  # (B,1,dim)
        se4 = tf.reduce_sum(out, axis=1, keepdims=True)  # (B,1,dim)
        #se5 = tf.reduce_prod(out, axis=1, keepdims=True)  # (B,1,dim)
        
        se = se1 + se2 + se3 + se4 #+ se5
        
        #se6 = se4 - (se1 + se2 + se3)
        
        #se = se6 #+ se6

        
        
        se = self.se_reduce(se)                          # (B,1,reduced_dim)
        se = self.se_expand(se)                          # (B,1,dim)
        
        #out = out * se

        #if self.use_scale:
         #   out *= self.global_scale
          #  se *= self.local_scale

        # Combine Global and Local Attention

        
        out2 = tf.sigmoid(se)
        
        
        return out * out2 #* se
        




# ----------------------------------------------------
# 2. MSWShiftedWindowDynamicFusion
# ----------------------------------------------------
import tensorflow as tf
from tensorflow.keras.layers import Layer, Dense, LayerNormalization
from tensorflow.signal import dct

class MSWShiftedWindowDynamicFusion(Layer):
    def __init__(self, dim, window_sizes=[7, 14], **kwargs):
        super().__init__(**kwargs)
        self.dim = dim
        self.window_sizes = window_sizes

        # one NODEAttentionLayer + norm per window size (assume NODEAttentionLayer defined)
        #self.attn_layers = [NODEAttentionLayer(dim) for _ in window_sizes]
        
        #self.attn_layers = [NODEAttentionLayer(dim) for _ in window_sizes]
        #self.norm_layers = [LayerNormalization(epsilon=1e-6) for _ in window_sizes]

        self.attn_layers = NODEAttentionLayer(dim)
        self.norm_layers = LayerNormalization(epsilon=1e-6)

        # frequency projection MLPs
        self.freq_proj = [
            Dense(dim, activation="sigmoid", name=f"freq_proj_{i}")
            for i in range(len(window_sizes))
        ]

        # spatial–frequency gating per scale
        self.sf_gates = [
            Dense(dim, activation="sigmoid", name=f"sf_gate_{i}")
            for i in range(len(window_sizes))
        ]

        # learnable base fusion weights
        self.alpha = self.add_weight(
            shape=(len(window_sizes),),
            initializer="ones",
            trainable=True,
            name="fusion_alpha"
        )

        self.weight1 = self.add_weight(shape=(1, 1, 1, 1),  initializer=tf.keras.initializers.HeNormal(), trainable=True, name='weight2', 
                                              constraint=tf.keras.constraints.MaxNorm(2.0))

    def call(self, x):
        B, H, W, C = tf.unstack(tf.shape(x))
        scale_outputs = []
        scale_energies = []

        for i, win in enumerate(self.window_sizes):
            shift = win // 4
            x_shift = tf.roll(x, shift=[shift, shift], axis=[1, 2])

            # pad so H,W divisible by win
            pad_h = (win - H % win) % win
            pad_w = (win - W % win) % win
            x_pad = tf.pad(x_shift, [[0,0],[0,pad_h],[0,pad_w],[0,0]])
            Hp, Wp = H + pad_h, W + pad_w

            # partition into windows: [B, nH, win, nW, win, C]
            nH, nW = Hp // win, Wp // win
            x_win = tf.reshape(x_pad, [B, nH, win, nW, win, C])

            # ---- 2D DCT along the two window dims ----
            # 1) DCT along window-height axis (original axis=2):
            perm1 = [0, 1, 3, 4, 5, 2]           # bring axis 2 to last
            x_p1 = tf.transpose(x_win, perm1)    # shape [B,nH,nW,win,C,win_h]
            d1 = dct(x_p1, type=2, norm="ortho") # DCT on last axis
            freq_h = tf.transpose(d1, [0,1,5,2,3,4])  # back to [B,nH,win,nW,win,C]

            # 2) DCT along window-width axis (original axis=4):
            perm2 = [0, 1, 2, 3, 5, 4]            # bring axis 4 to last
            x_p2 = tf.transpose(freq_h, perm2)   # [B,nH,win,nW,C,win_w]
            d2 = dct(x_p2, type=2, norm="ortho") # DCT on last axis
            freq_resh = tf.transpose(d2, [0,1,2,3,5,4])  # [B,nH,win,nW,win,C]

            # compute per-window energy
            
            #energy = tf.reduce_mean(tf.abs(freq_resh), axis=[2,4,5])  # [B, nH, nW]
            #scale_energies.append(tf.reduce_mean(energy, axis=[1,2])) # [B]

            abs_freq = tf.abs(freq_resh)

            energy_mean = tf.reduce_mean(abs_freq, axis=[2,4,5])  # [B, nH, nW]
            energy_sum  = tf.reduce_sum(abs_freq, axis=[2,4,5])  # [B, nH, nW]
            energy_max  = tf.reduce_max(abs_freq, axis=[2,4,5])  # [B, nH, nW]
            energy_min  = tf.reduce_min(abs_freq, axis=[2,4,5])  # [B, nH, nW]
            #energy_prod  = tf.reduce_prod(abs_freq, axis=[2,4,5])  # [B, nH, nW]
            
            # 2) now collapse across all windows (axes 1 & 2) → one value per batch
            scale_mean = tf.reduce_mean(energy_mean, axis=[1,2])  # [B]
            scale_sum  = tf.reduce_sum(energy_sum,  axis=[1,2])  # [B]
            scale_max  = tf.reduce_max(energy_max,  axis=[1,2])  # [B]
            scale_min  = tf.reduce_min(energy_min,  axis=[1,2])  # [B]
            #scale_prod  = tf.reduce_prod(energy_prod,  axis=[1,2])  # [B]

            scale_stats = scale_mean + scale_sum + scale_max + scale_min #+ scale_prod
            
            # 3) collect them however you like—for instance, stack into a single [B,4] vector:
            #scale_stats = tf.stack([scale_mean, scale_sum, scale_max, scale_min, scale_prod], axis=1)  # [B,4]
            scale_energies.append(scale_stats)

            
            #scale_energies.append(stat_mean)
            #scale_energies.append(stat_sum)
            #scale_energies.append(stat_max)
            #scale_energies.append(stat_min)
            
            
            
            # flatten windows → tokens
            seq = win * win
            freq_flat = tf.reshape(freq_resh, [B * nH * nW, seq, C])
            spat_flat = tf.reshape(x_win,   [B * nH * nW, seq, C])

            # freq → token-space
            fp = self.freq_proj[i](freq_flat)
            gate = self.sf_gates[i](spat_flat)
            fused = gate * fp + (1.0 - gate) * spat_flat

        
        # attention + reshape back
        attn_out = self.attn_layers(fused)
        attn_resh = tf.reshape(attn_out, [B, nH, nW, win, win, C])
        attn_resh = tf.reshape(attn_resh, [B, Hp, Wp, C])
        attn_unpad = attn_resh[:, :H, :W, :]
        attn_final = tf.roll(attn_unpad, shift=[-shift, -shift], axis=[1,2])

        # add & norm
        scale_outputs.append(self.norm_layers(attn_final + x))

        # fuse across scales
        
        energies = tf.stack(scale_energies, axis=-1)   # [B, S]
        e_w = tf.nn.softmax(energies, axis=-1)         # per-sample
        a_w = tf.nn.softmax(self.alpha)                # global
        w = a_w[None, :] * e_w
        
        w = w / tf.reduce_sum(w, axis=-1, keepdims=True)
        
        #w2 = w / tf.reduce_mean(w, axis=-1, keepdims=True)
        #w3 = w / tf.reduce_max(w, axis=-1, keepdims=True)
        #w4 = w / tf.reduce_min(w, axis=-1, keepdims=True)

        #w = tf.sigmoid(w1 + w2 + w3 + w4)
        
        #out1 = tf.sigmoid(out * se)

        
        stacked = tf.stack(scale_outputs, axis=-1)     # [B,H,W,C,S]
        out = tf.reduce_sum(stacked * w[:, None, None, None, :], axis=-1)
        
        enhanced = x + out #* self.weight1
        
        return enhanced
```

```python


def residual_GLC_branch1(inputs1, inputs2):
    
    x1 = Conv2D(filters=64, kernel_size=(7, 7), strides=(2, 2), padding='same')(inputs1)
    
    x1 = BatchNormalization()(x1)
    x1 = tf.keras.layers.Activation('relu')(x1)
    x1 = MaxPooling2D(pool_size=(3, 3), strides=(2, 2), padding='same')(x1)
    
    x2 = Conv2D(filters=64, kernel_size=(7, 7), strides=(2, 2), padding='same')(inputs2)
    
    x2 = BatchNormalization()(x2)
    x2 = tf.keras.layers.Activation('relu')(x2)
    x2 = MaxPooling2D(pool_size=(3, 3), strides=(2, 2), padding='same')(x2)
    

    x1 = RGSA(x1, filters=64)
    x1 = tf.keras.layers.Dropout(0.25)(x1, training = True)  ## MCD ####
    
    x2 = RGSA(x2, filters=64)
    x2 = tf.keras.layers.Dropout(0.25)(x2, training = True)  ## MCD ####

    
    x_1, x_2 = AttentionBlock(channels = 64)([x1, x2])  ## MIFA ####

    print('64 x_1, x_2 shape:',x_1.shape, x_2.shape)

    
    
    x1 = RGSA(x_1, filters=64)
    x1 = tf.keras.layers.Dropout(0.25)(x1, training = True)  ## MCD ####
    
    x2 = RGSA(x_2, filters=64)
    x2 = tf.keras.layers.Dropout(0.25)(x2, training = True)  ## MCD ####

    x1 = MSWShiftedWindowDynamicFusion(dim=64, window_sizes=[5,10])(x1)
    x2 = MSWShiftedWindowDynamicFusion(dim=64, window_sizes=[4,8])(x2)
    
    
    x_11, x_22 = AttentionBlock(channels = 64)([x1, x2])  ## MIFA ####
    print('x_11, x_22 shape:',x_11.shape, x_22.shape)


    
    
    #x11, x21 = info_fusion(x_11, x_22, 128)
    #x11, x21 = info_fusion(x11, x21, 256)
    #x11, x21 = info_fusion(x11, x21, 512)
    

    
    
    x1 = RGSA(x_11, filters=128, strides=(2, 2), use_projection=True)
    x1 = tf.keras.layers.Dropout(0.25)(x1, training = True)  ## MCD ####
    
    x2 = RGSA(x_22, filters=128, strides=(2, 2), use_projection=True)
    x2 = tf.keras.layers.Dropout(0.25)(x2, training = True)  ## MCD ####

    
    
    x_111, x_222 = AttentionBlock(channels = 128)([x1, x2])  ## MIFA ####
    print('x_111, x_222 shape:',x_111.shape, x_222.shape)
    
    x1 = RGSA(x_111, filters=128)
    x1 = tf.keras.layers.Dropout(0.25)(x1, training = True)  ## MCD ####
    
    x2 = RGSA(x_222, filters=128)
    x2 = tf.keras.layers.Dropout(0.25)(x2, training = True)  ## MCD ####

    
    x1 = MSWShiftedWindowDynamicFusion(dim=128, window_sizes=[5,10])(x1)
    x2 = MSWShiftedWindowDynamicFusion(dim=128, window_sizes=[4,8])(x2)
    
    x_1111, x_2222 = AttentionBlock(channels = 128)([x1, x2])  ## MIFA ####
    print('x_1111, x_2222 shape:',x_1111.shape, x_2222.shape)


    
    #x111, x211 = info_fusion(x_1111, x_2222, 256)
    #x111, x211 = info_fusion(x111, x211, 512)

    
    
    x1 = RGSA(x_1111, filters=256, strides=(2, 2), use_projection=True)
    x1 = tf.keras.layers.Dropout(0.25)(x1, training = True)  ## MCD ####
    
    x2 = RGSA(x_2222, filters=256, strides=(2, 2), use_projection=True)
    x2 = tf.keras.layers.Dropout(0.25)(x2, training = True)  ## MCD ####

    
    
    x_11111, x_22222 = AttentionBlock(channels = 256)([x1, x2])  ## MIFA ####
    print('x_11111, x_22222 shape:',x_11111.shape, x_22222.shape)
    
    
    x1 = RGSA(x_11111, filters=256)
    x1 = tf.keras.layers.Dropout(0.25)(x1, training = True)  ## MCD ####
    
    x2 = RGSA(x_22222, filters=256)
    x2 = tf.keras.layers.Dropout(0.25)(x2, training = True)  ## MCD ####

    x1 = MSWShiftedWindowDynamicFusion(dim=256, window_sizes=[5,10])(x1)
    x2 = MSWShiftedWindowDynamicFusion(dim=256, window_sizes=[4,8])(x2)
    
    
    x_12, x_21 = AttentionBlock(channels = 256)([x1, x2])  ## MIFA ####
    print('x_12, x_21 shape:',x_12.shape, x_21.shape)

    
    
    #x1111, x2111 = info_fusion(x_12, x_21, 512)
    

    
    x1 = RGSA(x_12, filters=512, strides=(2, 2), use_projection=True)
    
    
    x2 = RGSA(x_21, filters=512, strides=(2, 2), use_projection=True)

    
    
    x_112, x_211 = AttentionBlock(channels = 512)([x1, x2])  ## MIFA ####
    print('x_112, x_211 shape:',x_112.shape, x_211.shape)
    
    x1 = RGSA(x_112, filters=512)
    x2 = RGSA(x_211, filters=512)

    x1 = MSWShiftedWindowDynamicFusion(dim=512, window_sizes=[5,10])(x1)
    x2 = MSWShiftedWindowDynamicFusion(dim=512, window_sizes=[4,8])(x2)
    
    x_1112, x_2111 = AttentionBlock(channels = 512)([x1, x2])
    print('x_1112, x_2111 shape:',x_1112.shape, x_2111.shape)

    # x11, x21, x111, x211, x1111, x2111, x_1112, x_2111

    x__1 = tf.keras.layers.Concatenate(axis=-1)([x_1112, x_2111])
    x__1 = tf.keras.layers.Dropout(0.25)(x__1, training = True)  ## MCD ####
    
    return x__1
```
